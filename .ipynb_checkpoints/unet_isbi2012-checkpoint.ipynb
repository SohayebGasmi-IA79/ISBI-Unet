# Synthesis of *U-Net: Convolutional Networks for Biomedical Image Segmentation*
**Ronneberger, Fischer, Brox (2015), arXiv:1505.04597**

— Master IAA

This notebook re-implements the paper and runs it on the **real ISBI 2012 EM segmentation data** (30 training slices of a *Drosophila* ventral nerve cord, 512x512). The data is downloaded **from inside the notebook** (Kaggle through `kagglehub`, with a GitHub mirror as automatic fallback).

Contents: (1) data download, (2) weight map (Eq. 2), (3) elastic deformations (Sec. 3.1), (4) architecture bookkeeping, (5) overlap-tile, (6) He init, (7) U-Net in PyTorch, (8) training with weighted soft-max cross-entropy (Eq. 1), (9) results.



## 1. Setup and data download

```python
import os, re, sys, time, subprocess, urllib.request
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
from scipy import ndimage as ndi
from PIL import Image, ImageSequence

try:
    import kagglehub
except ImportError:
    subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', 'kagglehub'], check=True)
    import kagglehub

os.makedirs('figs', exist_ok=True)
np.random.seed(0)
plt.rcParams['figure.dpi'] = 110
```

```python
# ISBI 2012 EM segmentation challenge (train-volume / train-labels, 30 slices of 512x512)
KAGGLE_SLUGS = ['hamzamohiuddin/isbi-2012-challenge',      # https://www.kaggle.com/datasets/hamzamohiuddin/isbi-2012-challenge
                'purplezap/isbi-2012-challenge']           # https://www.kaggle.com/datasets/purplezap/isbi-2012-challenge
GITHUB_BASE = 'https://raw.githubusercontent.com/zhixuhao/unet/master/data/membrane/train'   # same data as 30 PNG pairs

def download_isbi():
    for slug in KAGGLE_SLUGS:
        try:
            p = kagglehub.dataset_download(slug)
            print('Kaggle download OK:', slug, '->', p)
            return Path(p)
        except Exception as e:
            print(f'Kaggle failed for {slug}: {type(e).__name__}: {str(e)[:150]}')
    print('Falling back to the GitHub mirror ...')
    root = Path('isbi2012_github')
    for kind in ('image', 'label'):
        (root / kind).mkdir(parents=True, exist_ok=True)
        for i in range(30):
            f = root / kind / f'{i}.png'
            if not f.exists():
                urllib.request.urlretrieve(f'{GITHUB_BASE}/{kind}/{i}.png', f)
    return root

DATA_ROOT = download_isbi()
for p in sorted(Path(DATA_ROOT).rglob('*'))[:15]:
    print('   ', p.relative_to(DATA_ROOT))
```

## 2. Loading, instances and the pixel-wise weight map (Eq. 2)
The labels are binary (cell interior vs. membrane). Membrane is the **minority class**, so it is detected automatically (robust to the label convention of each Kaggle copy). Cell *instances* are the connected components of the non-membrane class. Weight map of the paper:

$$w(x) = w_c(x) + w_0 \exp\left(-\frac{(d_1(x)+d_2(x))^2}{2\sigma^2}\right),\quad w_0 = 10,\ \sigma = 5\text{ px}$$

with $d_1, d_2$ the distances to the nearest and second-nearest cell and $w_c$ balancing class frequencies.

```python
def natural_key(p):
    return [int(t) if t.isdigit() else t for t in re.split(r'(\d+)', str(p).lower())]

def read_frames(p):                       # handles single images and multi-page TIFF stacks
    return [np.array(f.convert('L')) for f in ImageSequence.Iterator(Image.open(p))]

def load_isbi(root):
    root = Path(root)
    files = [p for p in root.rglob('*') if p.is_file() and p.suffix.lower() in ('.tif', '.tiff', '.png', '.jpg', '.jpeg')]
    files = [p for p in files if 'test' not in str(p.relative_to(root)).lower()]      # test labels are not public
    is_lab = lambda p: re.search(r'label|mask', str(p.relative_to(root)).lower()) is not None
    lab_f = sorted([p for p in files if is_lab(p)], key=natural_key)
    img_f = sorted([p for p in files if not is_lab(p)], key=natural_key)
    imgs = [a for p in img_f for a in read_frames(p)]
    labs = [a for p in lab_f for a in read_frames(p)]
    print(f'{len(img_f)} image file(s) -> {len(imgs)} images | {len(lab_f)} label file(s) -> {len(labs)} labels')
    assert len(imgs) == len(labs) > 0, 'could not pair images and labels: check the folder listing above'
    return np.stack(imgs).astype(np.float32) / 255., np.stack(labs)

def weight_map(inst, w0=10.0, sigma=5.0):
    mem = inst == 0
    d1 = np.full(inst.shape, np.inf, np.float32); d2 = d1.copy()
    for i in np.unique(inst[inst > 0]):                          # running two smallest distances
        d = ndi.distance_transform_edt(inst != i).astype(np.float32)
        d2 = np.minimum(d2, np.maximum(d1, d)); d1 = np.minimum(d1, d)
    f = mem.mean()                                               # membrane frequency
    wc = np.where(mem, 0.5 / f, 0.5 / (1 - f))                   # class-frequency balancing
    return (wc + w0 * np.exp(-((d1 + d2) ** 2) / (2 * sigma ** 2))).astype(np.float32)

X, L = load_isbi(DATA_ROOT)
mem = L > 127
if mem.mean() > 0.5:                      # membrane must be the minority class
    mem = ~mem
MEM = mem.astype(np.uint8)
print('images', X.shape, '| membrane ratio: %.3f' % MEM.mean())
assert 0.03 < MEM.mean() < 0.5 and min(X.shape[1:]) >= 388, 'unexpected labels or image size'

t0 = time.time()
INST = np.stack([ndi.label(m == 0)[0] for m in MEM]).astype(np.int32)     # cell instances
WMAP = np.stack([weight_map(i) for i in INST])
print('cells per image: %.0f (avg) | weight range %.2f - %.2f | %.0fs' % (
      np.mean([i.max() for i in INST]), WMAP.min(), WMAP.max(), time.time() - t0))

# split: 24 slices for training, 6 held out (the official test labels are not public).
# Note: slices come from one volume, so the last held-out slices are close to training slices.
N = len(X); val_idx = np.arange(N - 6, N); train_idx = np.arange(N - 6)
mu, sd = X[train_idx].mean(), X[train_idx].std()
Xn = ((X - mu) / sd).astype(np.float32)

fig, ax = plt.subplots(1, 4, figsize=(15, 4))
ax[0].imshow(X[0], cmap='gray'); ax[0].set_title('(a) EM image')
ax[1].imshow(MEM[0], cmap='gray'); ax[1].set_title('(b) membrane (class 1)')
ax[2].imshow(np.ma.masked_equal(INST[0], 0), cmap='tab20'); ax[2].set_facecolor('k'); ax[2].set_title('(c) cell instances')
im = ax[3].imshow(WMAP[0], cmap='jet'); ax[3].set_title('(d) weight map w(x)'); plt.colorbar(im, ax=ax[3], fraction=0.046)
for a in ax: a.axis('off')
plt.tight_layout(); plt.savefig('figs/fig_data_weights.png', bbox_inches='tight'); plt.show()
```

## 3. Elastic deformations (Section 3.1)
Random displacements on a coarse **3x3 grid**, Gaussian with **sigma = 10 px**, **bicubic** interpolation to per-pixel displacements. The same field is applied to image, mask and weight map.

```python
def elastic_field(shape, grid=3, sigma=10.0, rng=None):
    rng = rng or np.random.default_rng()
    H, W = shape
    disp = rng.normal(0, sigma, (2, grid, grid))
    g = np.meshgrid(np.linspace(0, grid - 1, H), np.linspace(0, grid - 1, W), indexing='ij')
    dy = ndi.map_coordinates(disp[0], g, order=3, mode='nearest')
    dx = ndi.map_coordinates(disp[1], g, order=3, mode='nearest')
    yy, xx = np.mgrid[:H, :W].astype(np.float32)
    return yy + dy, xx + dx

def elastic_apply(arr, coords, order):
    return ndi.map_coordinates(arr, coords, order=order, mode='reflect')

coords = elastic_field(X[0].shape, rng=np.random.default_rng(1))
fig, ax = plt.subplots(1, 4, figsize=(15, 4))
for a, im_, t in zip(ax, [X[0], elastic_apply(X[0], coords, 3), MEM[0], elastic_apply(MEM[0].astype(np.float32), coords, 0)],
                     ['original image', 'elastic deformation', 'original mask', 'deformed mask']):
    a.imshow(im_, cmap='gray'); a.set_title(t); a.axis('off')
plt.tight_layout(); plt.savefig('figs/fig_elastic.png', bbox_inches='tight'); plt.show()
```

## 4. Architecture bookkeeping (Section 2, Fig. 1)
Exact feature-map sizes, number of convolutional layers (paper: 23) and parameters for a 572x572 input.

```python
def unet_trace(size=572, base=64, n_classes=2, in_ch=1):
    c = [base * 2**i for i in range(5)]
    conv = lambda cin, cout, k=3: k*k*cin*cout + cout
    rows, params, h, cin = [], 0, size, in_ch
    rows.append(('input', in_ch, h, 0))
    for lvl in range(4):
        for j in range(2):
            h -= 2; p = conv(cin, c[lvl]); params += p; cin = c[lvl]
            rows.append((f'enc{lvl+1} conv3x3 #{j+1}', c[lvl], h, p))
        assert h % 2 == 0
        h //= 2; rows.append((f'enc{lvl+1} maxpool 2x2', c[lvl], h, 0))
    for j in range(2):
        h -= 2; p = conv(cin, c[4]); params += p; cin = c[4]
        rows.append((f'bottom conv3x3 #{j+1}', c[4], h, p))
    for lvl in [3, 2, 1, 0]:
        h *= 2; p = 4*cin*c[lvl] + c[lvl]; params += p
        rows.append((f'dec{lvl+1} up-conv 2x2', c[lvl], h, p))
        cin = 2 * c[lvl]
        for j in range(2):
            h -= 2; p = conv(cin, c[lvl]); params += p; cin = c[lvl]
            rows.append((f'dec{lvl+1} conv3x3 #{j+1}', c[lvl], h, p))
    p = conv(cin, n_classes, 1); params += p
    rows.append(('final conv1x1', n_classes, h, p))
    return rows, params

rows, total = unet_trace()
print(f"{'layer':24s}{'channels':>9s}{'size':>12s}{'params':>13s}")
for name, ch, h, p in rows:
    print(f'{name:24s}{ch:9d}{str(h)+"x"+str(h):>12s}{p:13,d}')
print('\nConvolutional layers (3x3 + up-conv + 1x1):', sum('conv' in r[0] for r in rows))
print(f'Total parameters: {total:,} | output for 572x572 input: {rows[-1][2]}x{rows[-1][2]}')
```

## 5. Overlap-tile strategy (Fig. 2) and He initialisation
Unpadded convolutions shrink a 572x572 tile to 388x388: **92 px** of context is lost per side. The image is **mirrored** at its borders and covered by 388x388 output tiles. Weights are initialised with $\mathcal N(0,\sqrt{2/N})$ (He).

```python
tile_in, tile_out = 572, 388
m = (tile_in - tile_out) // 2
H, W = X[0].shape
nty, ntx = int(np.ceil(H / tile_out)), int(np.ceil(W / tile_out))
padded = np.pad(X[0], ((m, m + nty*tile_out - H), (m, m + ntx*tile_out - W)), mode='reflect')
print(f'image {H}x{W} -> {nty}x{ntx} = {nty*ntx} tiles | margin {m}px | padded {padded.shape}')
fig, ax = plt.subplots(figsize=(6, 6)); ax.imshow(padded, cmap='gray')
ax.add_patch(plt.Rectangle((0, 0), tile_in, tile_in, fill=False, ec='royalblue', lw=2, label='input tile 572x572'))
ax.add_patch(plt.Rectangle((m, m), tile_out, tile_out, fill=False, ec='gold', lw=2, label='output tile 388x388'))
ax.add_patch(plt.Rectangle((m, m), W, H, fill=False, ec='red', ls='--', lw=1, label='original image'))
ax.legend(loc='lower right', fontsize=8); ax.set_title('Mirror padding + overlap-tile'); ax.axis('off')
plt.savefig('figs/fig_tiles.png', bbox_inches='tight'); plt.show()

# He initialisation: signal variance through 20 ReLU layers
N_in = 9 * 64
def forward_var(std, depth=20, width=N_in, seed=0):
    r = np.random.default_rng(seed); x = r.normal(size=(512, width))
    for _ in range(depth):
        x = np.maximum(x @ r.normal(0, std, (width, width)), 0)
    return x.var()
print('N = %d | He std = sqrt(2/N) = %.4f' % (N_in, np.sqrt(2 / N_in)))
print('variance after 20 ReLU layers, He init   : %.3e' % forward_var(np.sqrt(2 / N_in)))
print('variance after 20 ReLU layers, std = 0.01: %.3e' % forward_var(0.01))
```

## 6. PyTorch U-Net (valid convolutions, crop-and-concatenate, drop-out, He init)

```python
import torch, torch.nn as nn, torch.nn.functional as F
device = 'cuda' if torch.cuda.is_available() else 'cpu'
BASE = 64 if device == 'cuda' else 16          # reduced width on CPU only
print('device:', device, '| base channels:', BASE)

def double_conv(cin, cout):
    return nn.Sequential(nn.Conv2d(cin, cout, 3), nn.ReLU(inplace=True),
                         nn.Conv2d(cout, cout, 3), nn.ReLU(inplace=True))

def center_crop(t, hw):
    _, _, h, w = t.shape
    i, j = (h - hw[0]) // 2, (w - hw[1]) // 2
    return t[:, :, i:i + hw[0], j:j + hw[1]]

class UNet(nn.Module):
    def __init__(self, in_ch=1, n_classes=2, base=64, p_drop=0.5):
        super().__init__()
        c = [base * 2**i for i in range(5)]
        self.enc = nn.ModuleList([double_conv(in_ch, c[0])] + [double_conv(c[i], c[i+1]) for i in range(3)])
        self.pool = nn.MaxPool2d(2, 2)
        self.bottom = nn.Sequential(double_conv(c[3], c[4]), nn.Dropout2d(p_drop))   # drop-out at the end of the contracting path
        self.up  = nn.ModuleList([nn.ConvTranspose2d(c[i+1], c[i], 2, stride=2) for i in [3, 2, 1, 0]])
        self.dec = nn.ModuleList([double_conv(2 * c[i], c[i]) for i in [3, 2, 1, 0]])
        self.out = nn.Conv2d(c[0], n_classes, 1)
        for mod in self.modules():                                                   # He initialisation
            if isinstance(mod, (nn.Conv2d, nn.ConvTranspose2d)):
                nn.init.kaiming_normal_(mod.weight, nonlinearity='relu'); nn.init.zeros_(mod.bias)

    def forward(self, x):
        skips = []
        for e in self.enc:
            x = e(x); skips.append(x); x = self.pool(x)
        x = self.bottom(x)
        for up, dec, s in zip(self.up, self.dec, reversed(skips)):
            x = up(x)
            x = dec(torch.cat([center_crop(s, x.shape[2:]), x], dim=1))
        return self.out(x)

net = UNet(base=BASE).to(device)
with torch.no_grad():
    y = net.eval()(torch.zeros(1, 1, 572, 572, device=device))
print('output shape for 572x572 input:', tuple(y.shape), '| parameters: %s' % f'{sum(p.numel() for p in net.parameters()):,}')
del net
```

## 7. Training utilities: weighted soft-max cross-entropy (Eq. 1) and evaluation
Each step uses **one** 572x572 input tile (batch size 1) cut from a randomly deformed, flipped and rotated slice (mirror-padded), with the pixel-wise weights of the matching 388x388 output window. Metrics on the held-out slices (tiled inference on the whole 512x512 slice):
* **IoU** of the membrane class and **pixel error**,
* **Rand index** between predicted and true *cell instances* (connected components of the non-membrane prediction), computed on true cell pixels: it penalises merged touching cells, which is precisely what the weight map targets.

```python
IN, OUT = 572, 388
M = (IN - OUT) // 2

def sample_tile(im, mk, w, rng, elastic=True):
    if elastic:
        coords = elastic_field(im.shape, rng=rng)
        im = elastic_apply(im, coords, 3); mk = elastic_apply(mk.astype(np.float32), coords, 0); w = elastic_apply(w, coords, 1)
    k, f = int(rng.integers(4)), int(rng.integers(2))
    tf = lambda a: np.rot90(a[:, ::-1] if f else a, k)
    im, mk, w = tf(im), tf(mk), tf(w)
    H, W = im.shape
    oy, ox = int(rng.integers(0, H - OUT + 1)), int(rng.integers(0, W - OUT + 1))
    imp = np.pad(im, M, mode='reflect')
    x  = imp[oy:oy + IN, ox:ox + IN]            # input tile centred on the output window
    y  = (mk[oy:oy + OUT, ox:ox + OUT] > 0.5)
    ww = w[oy:oy + OUT, ox:ox + OUT]
    return np.ascontiguousarray(x), np.ascontiguousarray(y), np.ascontiguousarray(ww)

def weighted_ce(logits, target, weight):
    nll = -F.log_softmax(logits, 1).gather(1, target[:, None]).squeeze(1)
    return (weight * nll).sum() / weight.sum()

@torch.no_grad()
def predict_prob(model, img, tile_out=OUT, tile_in=IN):
    m = (tile_in - tile_out) // 2
    H, W = img.shape
    Hp, Wp = int(np.ceil(H / tile_out)) * tile_out, int(np.ceil(W / tile_out)) * tile_out
    pad = np.pad(img, ((m, m + Hp - H), (m, m + Wp - W)), mode='reflect')
    out = np.zeros((Hp, Wp), np.float32); model.eval()
    for y in range(0, Hp, tile_out):
        for x in range(0, Wp, tile_out):
            t = torch.from_numpy(np.ascontiguousarray(pad[y:y + tile_in, x:x + tile_in]))[None, None].to(device)
            out[y:y + tile_out, x:x + tile_out] = torch.softmax(model(t), 1)[0, 1].cpu().numpy()
    return out[:H, :W]

def rand_index(a, b):
    a = a.ravel().astype(np.int64); b = b.ravel().astype(np.int64); n = a.size
    _, ai = np.unique(a, return_inverse=True); _, bi = np.unique(b, return_inverse=True)
    cont = np.bincount(ai * (bi.max() + 1) + bi)
    c2 = lambda v: (v.astype(np.float64) * (v - 1) / 2).sum()
    tot = n * (n - 1) / 2
    return (tot + 2 * c2(cont) - c2(np.bincount(ai)) - c2(np.bincount(bi))) / tot

def metrics(prob, gt_mem, gt_inst):
    pred = prob > 0.5; gt = gt_mem == 1
    iou = (pred & gt).sum() / max((pred | gt).sum(), 1)
    pe = (pred != gt).mean()
    pred_inst = ndi.label(~pred)[0]
    cell = gt_inst > 0
    return iou, pe, rand_index(gt_inst[cell], pred_inst[cell])

def evaluate(model):
    r = np.array([metrics(predict_prob(model, Xn[i]), MEM[i], INST[i]) for i in val_idx])
    return r.mean(0)          # IoU, pixel error, Rand index

EPOCHS = 150 if device == 'cuda' else 2   # 150 epochs x 24 tiles = 3600 SGD steps (~10 min on a T4)
LR = 1e-3                     # with momentum 0.99 (effective step ~ 100x larger)

def train(use_weights=True, use_elastic=True, epochs=EPOCHS, seed=0, tag='full'):
    torch.manual_seed(seed); rng = np.random.default_rng(seed)
    model = UNet(base=BASE).to(device)
    opt = torch.optim.SGD(model.parameters(), lr=LR, momentum=0.99)
    hist = {'loss': [], 'iou': [], 'pe': [], 'rand': []}; t0 = time.time()
    for ep in range(epochs):
        model.train(); tot = 0
        for idx in rng.permutation(train_idx):
            x, y, w = sample_tile(Xn[idx], MEM[idx], WMAP[idx], rng, elastic=use_elastic)
            if not use_weights: w = np.ones_like(w)
            x = torch.from_numpy(x)[None, None].to(device)
            y = torch.from_numpy(y).long()[None].to(device)
            w = torch.from_numpy(w)[None].to(device)
            opt.zero_grad(); loss = weighted_ce(model(x), y, w); loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), 5.0)     # safety net, not in the paper
            opt.step(); tot += loss.item()
        iou, pe, ri = evaluate(model)
        hist['loss'].append(tot / len(train_idx)); hist['iou'].append(iou); hist['pe'].append(pe); hist['rand'].append(ri)
        print(f'[{tag}] epoch {ep+1:02d}/{epochs} | loss {hist["loss"][-1]:.4f} | IoU {iou:.4f} | pixel err {pe:.4f} | Rand {ri:.4f} | {time.time()-t0:.0f}s')
    return model, hist
```

## 8. Training the full model (weight map + elastic deformations)

```python
model, hist_full = train(tag='weighted U-Net')
```

## 9. Results: learning curves, predictions and instance separation

```python
fig, ax = plt.subplots(1, 3, figsize=(15, 3.6))
ax[0].plot(hist_full['loss'], 'o-'); ax[0].set_title('Weighted training loss')
ax[1].plot(hist_full['iou'], 'o-', c='tab:green'); ax[1].set_title('Held-out IoU (membrane)')
ax[2].plot(hist_full['rand'], 'o-', c='tab:red'); ax[2].set_title('Held-out Rand index (cell instances)')
for a in ax: a.set_xlabel('epoch')
plt.tight_layout(); plt.savefig('figs/curves.png', bbox_inches='tight'); plt.show()

i = val_idx[0]
prob = predict_prob(model, Xn[i]); pred = prob > 0.5
iou, pe, ri = metrics(prob, MEM[i], INST[i])
print('slice %d | IoU %.4f | pixel error %.4f | Rand %.4f' % (i, iou, pe, ri))
fig, ax = plt.subplots(2, 3, figsize=(14, 9))
ax[0, 0].imshow(X[i], cmap='gray'); ax[0, 0].set_title('input')
ax[0, 1].imshow(MEM[i], cmap='gray'); ax[0, 1].set_title('ground truth (membrane)')
ax[0, 2].imshow(pred, cmap='gray'); ax[0, 2].set_title('U-Net prediction (overlap-tile)')
ax[1, 0].imshow(prob, cmap='viridis'); ax[1, 0].set_title('membrane probability')
ax[1, 1].imshow(np.ma.masked_equal(INST[i], 0), cmap='tab20'); ax[1, 1].set_facecolor('k'); ax[1, 1].set_title('true cell instances')
ax[1, 2].imshow(np.ma.masked_equal(ndi.label(~pred)[0], 0), cmap='tab20'); ax[1, 2].set_facecolor('k'); ax[1, 2].set_title('predicted cell instances')
for a in ax.ravel(): a.axis('off')
plt.tight_layout(); plt.savefig('figs/prediction.png', bbox_inches='tight'); plt.show()
```

## 10. Conclusion
The notebook reproduces the four ideas that make U-Net work with very few annotated images: (i) the symmetric contracting/expanding architecture with skip connections, (ii) heavy elastic data augmentation, (iii) the border-aware weighted loss, and (iv) seamless overlap-tile inference with mirrored borders. It is evaluated on the real ISBI 2012 slices (24 train / 6 held out); the official test labels are not public, so the numbers are **not** directly comparable with the paper's leaderboard results (warping error 0.000353, Rand error 0.0382).
